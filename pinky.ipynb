```python
# =============================================================================
# Trans-VaTabNet — Final Accuracy Boost + Shared/Common DEG Intersection
#   Targets 3-class ≥ 0.85 via multi-seed hierarchy + calibration + TTA + blend
# =============================================================================

# ------------------------- 1. Install packages -------------------------------
!pip install --upgrade pip -q
!pip install pandas numpy scikit-learn -q
!pip install torch torchvision --index-url https://download.pytorch.org/whl/cu121 -q
!pip install pytorch-tabnet -q
!pip install GEOparse imbalanced-learn gseapy mygene -q
!pip install rpy2 -q
!pip install shap matplotlib -q
!pip install xgboost -q

!apt-get update -qq
!apt-get install -y r-base r-base-dev libcurl4-openssl-dev libssl-dev \
                    libxml2-dev libfontconfig1-dev libharfbuzz-dev \
                    libfribidi-dev libfreetype6-dev libpng-dev \
                    libtiff5-dev libjpeg-dev > /dev/null 2>&1

import rpy2.robjects as ro
print("Installing R packages (edgeR, limma, locfit)… may take 3–5 min.")
ro.r('''
dir.create(Sys.getenv("R_LIBS_USER"), recursive = TRUE, showWarnings = FALSE)
.libPaths(c(Sys.getenv("R_LIBS_USER"), .libPaths()))
if (!requireNamespace("BiocManager", quietly = TRUE)) {
    install.packages("BiocManager", repos = "https://cloud.r-project.org", quiet = TRUE)
}
pkgs <- c("edgeR", "limma", "locfit")
missing <- pkgs[!vapply(pkgs, requireNamespace, logical(1), quietly = TRUE)]
if (length(missing) > 0) BiocManager::install(missing, ask = FALSE, update = FALSE, quiet = TRUE)
suppressMessages({library(edgeR); library(limma)})
''')
print("All R packages installed and loaded.\n")

# ------------------------- 2. Imports ---------------------------------------
import os, random, warnings, contextlib, io

os.environ['CUDA_VISIBLE_DEVICES'] = '0,1'
warnings.filterwarnings('ignore')

import torch
print(f"torch {torch.__version__} | CUDA: {torch.cuda.is_available()}")
torch.manual_seed(42)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(42)

import numpy as np
np.random.seed(42)
random.seed(42)

import pandas as pd
import re
import GEOparse
import gseapy as gp
import mygene
import shap
import matplotlib.pyplot as plt

from scipy.stats import median_abs_deviation
from scipy.optimize import minimize
from scipy.special import softmax

from sklearn.model_selection import (StratifiedKFold,
                                     RepeatedStratifiedKFold)
from sklearn.preprocessing import LabelEncoder, RobustScaler, StandardScaler
from sklearn.feature_selection import VarianceThreshold, SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import (roc_auc_score, accuracy_score,
                             f1_score, balanced_accuracy_score,
                             matthews_corrcoef, confusion_matrix)
from pytorch_tabnet.tab_model import TabNetClassifier
from imblearn.over_sampling import SMOTE
from xgboost import XGBClassifier

import rpy2.robjects as ro
from rpy2.robjects import numpy2ri, pandas2ri
from rpy2.robjects.conversion import localconverter

numpy2ri.activate()
pandas2ri.activate()

import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

if torch.cuda.is_available():
    device = torch.device('cuda:0')
    torch.backends.cudnn.benchmark = True
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True
    TABNET_DEVICE = 'cuda'
    print(f"Using device: {device} ({torch.cuda.get_device_name(0)})\n")
else:
    device = torch.device('cpu')
    TABNET_DEVICE = 'cpu'
    print("Using device: CPU\n")

# =============================================================================
# HYPERPARAMETERS
# =============================================================================
LATENT_DIM       = 64
HIDDEN_DIM       = 256
VAE_EPOCHS       = 200
TABNET_EPOCHS    = 150
TABNET_N_D       = 32
TABNET_N_A       = 32
TABNET_N_STEPS   = 4
TABNET_GAMMA     = 1.5
TABNET_PATIENCE  = 30
BETA_KL          = 2.0
CE_WEIGHT        = 2.0
VAE_LR           = 5e-4
VAE_WEIGHT_DECAY = 1e-4
DROPOUT          = 0.30
N_SPLITS         = 10
TABNET_SEEDS     = [42, 1337, 2024, 7, 99, 11, 22, 33, 55, 77, 101, 202]  # 12
TOP_N_GENES      = 500

VAE_BATCH_SIZE   = 32 if torch.cuda.is_available() else 16
TABNET_BATCH     = 32 if torch.cuda.is_available() else 16
TABNET_VBATCH    = 16 if torch.cuda.is_available() else 8

USE_ORDINAL_LOSS = True
ORDINAL_ALPHA    = 0.5
USE_FOCAL_LOSS   = True
FOCAL_GAMMA      = 1.5

# ---- Boost switches ----
TTA_ROUNDS       = 5
N_EL_SEEDS       = 5
USE_CALIBRATION  = True
USE_SOFT_BLEND   = True
N_STACK_REPEATS  = 5
STACK_META_K     = 100

print(f"[HP] latent={LATENT_DIM}, hidden={HIDDEN_DIM}, dropout={DROPOUT}")
print(f"     seeds={len(TABNET_SEEDS)}, TTA={TTA_ROUNDS}, "
      f"EL_seeds={N_EL_SEEDS}")
print(f"     calibration={USE_CALIBRATION}, soft_blend={USE_SOFT_BLEND}\n")

# ------------------------- 3. Load data -------------------------------------
file_path = "/kaggle/input/datasets/srishtii02/gse216281-counts/GSE216281_counts.txt"

with open(file_path, 'r') as f:
    header_line = f.readline().strip()
    sample_names = [name.strip('"') for name in header_line.split()]
print(f"Number of samples: {len(sample_names)}")

gene_ids, counts_list = [], []
with open(file_path, 'r') as f:
    next(f)
    for line in f:
        parts = line.strip().split()
        if not parts: continue
        gene_id = parts[0].strip('"')
        counts = parts[1:]
        if len(counts) > len(sample_names): counts = counts[:len(sample_names)]
        elif len(counts) < len(sample_names):
            counts += [np.nan] * (len(sample_names) - len(counts))
        gene_ids.append(gene_id)
        counts_list.append(counts)

counts = pd.DataFrame(counts_list, index=gene_ids, columns=sample_names)
counts = counts.apply(pd.to_numeric, errors='coerce')
print(f"Count matrix shape: {counts.shape}\n")

print("Downloading GEO metadata...")
gse = GEOparse.get_GEO(geo="GSE216281", destdir="./")
phenotype = gse.phenotype_data

def extract_sample_name(title):
    m = re.search(r'sample_R\d+', title)
    return m.group(0) if m else None

phenotype['sample_name'] = phenotype['title'].apply(extract_sample_name)
braak_col = 'characteristics_ch1.2.braak lewy_body_stage'

def parse_braak_number(val):
    if not isinstance(val, str): return None
    try: return int(val.strip())
    except: return None

phenotype['braak_number'] = phenotype[braak_col].apply(parse_braak_number)

def braak_to_binary(b):
    if b is None: return None
    if b == 0: return 'Control'
    if 1 <= b <= 6: return 'PD'
    return None

def braak_to_stage(b):
    if b is None: return None
    if b == 0: return 'Control'
    if 1 <= b <= 4: return 'Early'
    if 5 <= b <= 6: return 'Late'
    return None

phenotype['group_binary'] = phenotype['braak_number'].apply(braak_to_binary)
phenotype['group_stage']  = phenotype['braak_number'].apply(braak_to_stage)

sample_to_binary, sample_to_stage = {}, {}
for _, row in phenotype.iterrows():
    sn = row['sample_name']
    if sn is None: continue
    if pd.notna(row['group_binary']): sample_to_binary[sn] = row['group_binary']
    if pd.notna(row['group_stage']):  sample_to_stage[sn]  = row['group_stage']

our_sample_names = counts.columns.tolist()
binary_labels = pd.Series([sample_to_binary.get(n, None) for n in our_sample_names],
                          index=our_sample_names, name='binary')
stage_labels = pd.Series([sample_to_stage.get(n, None) for n in our_sample_names],
                         index=our_sample_names, name='stage')

valid = binary_labels.notna() & stage_labels.notna()
counts = counts.loc[:, valid]
binary_labels = binary_labels[valid]
stage_labels  = stage_labels[valid]
print(f"After filtering: counts shape = {counts.shape}")

# ------------------------- 4. QC + outlier removal --------------------------
lib_sizes_raw = counts.sum(axis=0)
log_cpm_qc = np.log2(counts.div(lib_sizes_raw, axis=1) * 1e6 + 1)

def detect_outliers(log_cpm_df, n_mads=3.0, min_keep_frac=0.7):
    med = log_cpm_df.median(axis=1)
    mad = median_abs_deviation(log_cpm_df, axis=1, nan_policy='omit')
    z   = (log_cpm_df.sub(med, axis=0)).div(mad + 1e-8, axis=0)
    frac_out = (z.abs() > n_mads).mean(axis=0)
    keep = frac_out < (1 - min_keep_frac)
    if keep.sum() < len(keep) * 0.7:
        keep = frac_out <= frac_out.quantile(0.7)
    return keep

keep_mask = detect_outliers(log_cpm_qc, n_mads=3.0, min_keep_frac=0.7)
counts        = counts.loc[:, keep_mask.values]
binary_labels = binary_labels[keep_mask.values]
stage_labels  = stage_labels[keep_mask.values]
print(f"Outlier removal ENABLED — kept {keep_mask.sum()}/{len(keep_mask)} samples\n")

# ------------------------- 5. Filtering ------------------------------------
min_counts = 10
min_samples = int(0.1 * counts.shape[1])
keep = (counts >= min_counts).sum(axis=1) >= min_samples
filtered_counts = counts[keep]
print(f"After gene filtering: {filtered_counts.shape[0]} genes kept")

# ------------------------- 6. TMM + voom + Limma ---------------------------
print("\nRunning TMM + voom + Limma-voom in R...")
counts_mat     = filtered_counts.values.astype(float)
gene_names_r   = filtered_counts.index.tolist()
sample_names_r = filtered_counts.columns.tolist()
group_vec_r    = stage_labels.loc[filtered_counts.columns].tolist()

ro.globalenv['counts_mat']     = numpy2ri.numpy2rpy(counts_mat)
ro.globalenv['gene_names_r']   = ro.StrVector(gene_names_r)
ro.globalenv['sample_names_r'] = ro.StrVector(sample_names_r)
ro.globalenv['group_vec_r']    = ro.StrVector(group_vec_r)

ro.r('''
suppressMessages({library(edgeR); library(limma)})
rownames(counts_mat) <- gene_names_r
colnames(counts_mat) <- sample_names_r
group_factor <- factor(group_vec_r, levels = c("Control", "Early", "Late"))
dge <- DGEList(counts = counts_mat, group = group_factor)
dge <- normLibSizes(dge, method = "TMM")
design <- model.matrix(~ 0 + group_factor)
colnames(design) <- levels(group_factor)
v <- voom(dge, design, plot = FALSE)
voom_E <- as.matrix(v$E)
fit <- lmFit(v, design)
contrasts_mat <- makeContrasts(
    Early_vs_Control = Early - Control,
    Late_vs_Control  = Late - Control,
    Late_vs_Early    = Late - Early,
    levels = design
)
fit2 <- contrasts.fit(fit, contrasts_mat)
fit2 <- eBayes(fit2)
early_res <- topTable(fit2, coef = "Early_vs_Control", adjust = "BH", number = Inf)
late_res  <- topTable(fit2, coef = "Late_vs_Control",  adjust = "BH", number = Inf)
late_early_res <- topTable(fit2, coef = "Late_vs_Early", adjust = "BH", number = Inf)
''')

def r_df_to_pandas(r_df):
    with localconverter(ro.default_converter + pandas2ri.converter):
        df = ro.conversion.rpy2py(r_df)
    if not isinstance(df, pd.DataFrame): df = pd.DataFrame(df)
    return df

voom_expr      = pd.DataFrame(np.array(ro.globalenv['voom_E']),
                              index=list(ro.r('rownames(voom_E)')),
                              columns=list(ro.r('colnames(voom_E)')))
early_res      = r_df_to_pandas(ro.globalenv['early_res'])
late_res       = r_df_to_pandas(ro.globalenv['late_res'])
late_early_res = r_df_to_pandas(ro.globalenv['late_early_res'])
print(f"voom matrix: {voom_expr.shape}")

# ------------------------- 7. DEGs + Shared/Common intersection -------------
adj_p_thresh = 0.05
logfc_thresh = 1.0

def sig_set(res):
    return set(res[(res['adj.P.Val'] < adj_p_thresh) &
                   (res['logFC'].abs() > logfc_thresh)].index)

early_sig = sig_set(early_res)
late_sig  = sig_set(late_res)
le_sig    = sig_set(late_early_res)

# ---------- Shared / common DEG intersections ----------
shared_sig      = early_sig & late_sig                    # both early & late vs control
shared_all      = early_sig & late_sig & le_sig           # shared across all 3 contrasts
early_specific  = early_sig - late_sig
late_specific   = late_sig  - early_sig
early_only      = early_sig - late_sig - le_sig           # Early-only, not in Late or L-vs-E
late_only       = late_sig  - early_sig - le_sig          # Late-only, not in Early or L-vs-E

all_sig         = early_sig | late_sig | le_sig
two_of_three    = ((early_sig & late_sig) |
                   (early_sig & le_sig)  |
                   (late_sig  & le_sig)) - shared_all

print(f"Early-specific DEGs        : {len(early_specific)}")
print(f"Late-specific DEGs         : {len(late_specific)}")
print(f"Late-vs-Early DEGs         : {len(le_sig)}")
print(f"Shared (Early ∩ Late) DEGs : {len(shared_sig)}")
print(f"Shared in ALL 3 contrasts  : {len(shared_all)}")
print(f"Early-only DEGs            : {len(early_only)}")
print(f"Late-only DEGs             : {len(late_only)}")
print(f"Significant in 2-of-3      : {len(two_of_three)}")
print(f"Union of all DEGs          : {len(all_sig)}")

# Persist shared-DEG tables (sorted by combined significance)
shared_df = pd.DataFrame({
    'early_logFC':   early_res.reindex(sorted(shared_sig))['logFC'],
    'early_adjP':    early_res.reindex(sorted(shared_sig))['adj.P.Val'],
    'late_logFC':    late_res.reindex(sorted(shared_sig))['logFC'],
    'late_adjP':     late_res.reindex(sorted(shared_sig))['adj.P.Val'],
})
shared_df['max_adjP']   = shared_df[['early_adjP', 'late_adjP']].max(axis=1)
shared_df['mean_absFC'] = shared_df[['early_logFC', 'late_logFC']].abs().mean(axis=1)
shared_df = shared_df.sort_values('max_adjP')
shared_df.to_csv('shared_DEGs_early_and_late.csv')
print(f"Saved shared_DEGs_early_and_late.csv ({len(shared_df)} genes)")

mg = mygene.MyGeneInfo()
_GENE_CACHE = {}

def ensg_to_symbol_map(ensg_list):
    ensg_list = list(dict.fromkeys([g for g in ensg_list if g]))
    todo = [g for g in ensg_list if g not in _GENE_CACHE]
    if todo:
        cleaned = [g.split('.')[0] for g in todo]
        try:
            res = mg.querymany(cleaned, scopes='ensembl.gene', fields='symbol',
                               species='human', verbose=False, returnall=False)
        except Exception as e:
            print(f"[mygene] query failed: {e}")
            res = []
        cmap = {c: o for c, o in zip(cleaned, todo)}
        for item in res:
            q = item.get('query', '')
            orig = cmap.get(q, q)
            _GENE_CACHE[orig] = item.get('symbol', None)
    return {g: _GENE_CACHE.get(g) for g in ensg_list if _GENE_CACHE.get(g)}

# ------------------------- 8. GO / KEGG (incl. shared) ----------------------
print("\nRunning GO / KEGG enrichment on stage-specific AND shared DEGs…")

def safe_enrichr(gene_list, gene_sets):
    if len(gene_list) < 5: return None
    try:
        return gp.enrichr(gene_list=gene_list, organism='human',
                          gene_sets=gene_sets, cutoff=0.5, outdir=None)
    except Exception as e:
        print(f"  [enrichr] {gene_sets} failed: {e}")
        return None

all_ensg = list(early_specific | late_specific | shared_sig | shared_all)
ensg_map = ensg_to_symbol_map(all_ensg)

def to_symbols(ensg_set):
    return [ensg_map[g] for g in ensg_set if g in ensg_map]

gene_sets_to_query = [
    ('Early',      to_symbols(early_specific)),
    ('Late',       to_symbols(late_specific)),
    ('Shared',     to_symbols(shared_sig)),
    ('SharedAll3', to_symbols(shared_all)),
    ('EarlyOnly',  to_symbols(early_only)),
    ('LateOnly',   to_symbols(late_only)),
]

enrich_results = {}
for tag, glist in gene_sets_to_query:
    if len(glist) < 5:
        print(f"\n  [{tag}] skipped — only {len(glist)} mapped genes")
        continue
    for lib in ['GO_Biological_Process_2023', 'KEGG_2021_Human']:
        res = safe_enrichr(glist, lib)
        enrich_results[f'{tag}::{lib}'] = res
        if res is not None and hasattr(res, 'results') and res.results is not None:
            top = res.results.head(5)[['Term', 'Adjusted P-value']]
            print(f"\n  [{tag} / {lib}] Top-5 terms:")
            print(top.to_string(index=False))

# ------------------------- 9. Feature matrices ------------------------------
X_genes = voom_expr.T

target_ensg = [
    'ENSG00000204389','ENSG00000204388','ENSG00000188906',
    'ENSG00000132002','ENSG00000171862','ENSG00000179915',
    'ENSG00000251322','ENSG00000173110','ENSG00000109971',
]
biomarker_symbols = ['HSPA1A','HSPA1B','LRRK2','DNAJB1','PTEN',
                     'NRXN1','SHANK3','HSPA6','HSPA8']

vt = VarianceThreshold(threshold=0.01)
X_var = pd.DataFrame(vt.fit_transform(X_genes), index=X_genes.index,
                     columns=X_genes.columns[vt.get_support()])

le_bin = LabelEncoder(); y_bin = le_bin.fit_transform(binary_labels.loc[X_var.index].values)
le_3   = LabelEncoder(); y_3   = le_3.fit_transform(stage_labels.loc[X_var.index].values)

early_p = early_res['adj.P.Val']
late_p  = late_res['adj.P.Val']
le_p    = late_early_res['adj.P.Val']

# feature set A (binary)
de_A = pd.concat([early_p, late_p], axis=1).min(axis=1).sort_values().head(TOP_N_GENES).index
de_A_avail = [g for g in de_A if g in X_var.columns]
bm_avail   = [g for g in target_ensg if g in X_var.columns]
feature_set_A = list(dict.fromkeys(de_A_avail + bm_avail))
X_A = X_var[feature_set_A].values.astype(np.float32)

# feature set 3 (adds L-vs-E + shared-DEG signal)
de_3 = pd.concat([early_p, late_p, le_p], axis=1).min(axis=1).sort_values().head(TOP_N_GENES).index
de_3_avail       = [g for g in de_3 if g in X_var.columns]
early_spec_avail = [g for g in early_res.index if g in early_specific and g in X_var.columns]
late_spec_avail  = [g for g in late_res.index  if g in late_specific  and g in X_var.columns]
le_avail         = [g for g in late_early_res.index if g in le_sig and g in X_var.columns]
shared_avail     = [g for g in sorted(shared_sig) if g in X_var.columns]
shared_all_avail = [g for g in sorted(shared_all) if g in X_var.columns]

feature_set_3 = list(dict.fromkeys(
    de_3_avail + bm_avail
    + early_spec_avail[:200] + late_spec_avail[:150] + le_avail[:200]
    + shared_avail[:200] + shared_all_avail[:100]
))
X_3 = X_var[feature_set_3].values.astype(np.float32)

print(f"\nFeature set A (binary)   : {X_A.shape[0]} × {X_A.shape[1]}")
print(f"Feature set 3-class      : {X_3.shape[0]} × {X_3.shape[1]}")
print(f"  - shared DEG features  : {len(shared_avail[:200])}")
print(f"  - shared-all-3 features: {len(shared_all_avail[:100])}")
found_syms = [sym for sym, g in zip(biomarker_symbols, target_ensg) if g in X_var.columns]
print(f"Biomarkers found         : {len(found_syms)}/{len(target_ensg)} → {found_syms}")

# Binary sub-task views
stage_vals = stage_labels.loc[X_var.index].values
mask_late  = (stage_vals != 'Early')
y_late     = (stage_vals[mask_late] == 'Late').astype(int)
X_A_late   = X_A[mask_late]

mask_early = (stage_vals != 'Late')
y_early    = (stage_vals[mask_early] == 'Early').astype(int)
X_A_early  = X_A[mask_early]

# Early-vs-Late view
mask_pd = (stage_vals != 'Control')
X_EL    = X_3[mask_pd]
y_EL    = (stage_vals[mask_pd] == 'Late').astype(int)
print(f"Early-vs-Late subset     : {X_EL.shape[0]} samples "
      f"({(y_EL==0).sum()} Early, {(y_EL==1).sum()} Late)")

# =============================================================================
# VAE + TABNET
# =============================================================================
class SupervisedBetaVAE(nn.Module):
    def __init__(self, input_dim, num_classes,
                 hidden_dim=HIDDEN_DIM, latent_dim=LATENT_DIM, dropout=DROPOUT):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.LeakyReLU(0.1),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.BatchNorm1d(hidden_dim // 2),
            nn.LeakyReLU(0.1),
        )
        self.fc_mu      = nn.Linear(hidden_dim // 2, latent_dim)
        self.fc_logvar  = nn.Linear(hidden_dim // 2, latent_dim)
        self.classifier = nn.Sequential(
            nn.Linear(latent_dim, latent_dim),
            nn.LeakyReLU(0.1),
            nn.Dropout(dropout),
            nn.Linear(latent_dim, num_classes),
        )
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim // 2),
            nn.BatchNorm1d(hidden_dim // 2),
            nn.LeakyReLU(0.1),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim // 2, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.LeakyReLU(0.1),
            nn.Linear(hidden_dim, input_dim),
        )

    def encode(self, x):
        h = self.encoder(x)
        return self.fc_mu(h), self.fc_logvar(h)

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        return mu + torch.randn_like(std) * std

    def decode(self, z):     return self.decoder(z)
    def classify(self, z):   return self.classifier(z)

    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        x_recon = self.decode(z)
        logits  = self.classifier(mu)
        return x_recon, mu, logvar, z, logits


def svae_loss(x_recon, x, mu, logvar, logits, y,
              beta=BETA_KL, ce_weight=CE_WEIGHT, num_classes=3):
    recon = F.mse_loss(x_recon, x, reduction='sum') / x.size(0)
    kl    = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp()) / x.size(0)

    if USE_FOCAL_LOSS:
        ce_vec = F.cross_entropy(logits, y, reduction='none')
        pt     = torch.exp(-ce_vec)
        ce     = ((1 - pt) ** FOCAL_GAMMA * ce_vec).mean()
    else:
        ce = F.cross_entropy(logits, y)

    loss = recon + beta * kl + ce_weight * ce
    if USE_ORDINAL_LOSS and num_classes > 2:
        probs     = F.softmax(logits, dim=1)
        arange    = torch.arange(num_classes, device=logits.device).float()
        exp_class = (probs * arange).sum(dim=1)
        ord_pen   = F.mse_loss(exp_class, y.float())
        loss = loss + ORDINAL_ALPHA * ord_pen
    return loss


def train_supervised_beta_vae(X_train, y_train, num_classes,
                              hidden_dim=HIDDEN_DIM, latent_dim=LATENT_DIM,
                              epochs=VAE_EPOCHS, batch_size=VAE_BATCH_SIZE,
                              lr=VAE_LR, weight_decay=VAE_WEIGHT_DECAY,
                              beta=BETA_KL, ce_weight=CE_WEIGHT,
                              seed=42):
    input_dim = X_train.shape[1]
    torch.manual_seed(seed); np.random.seed(seed)
    if torch.cuda.is_available(): torch.cuda.manual_seed_all(seed)

    model = SupervisedBetaVAE(input_dim, num_classes,
                              hidden_dim=hidden_dim,
                              latent_dim=latent_dim).to(device)
    opt   = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)

    X_t = torch.tensor(X_train, dtype=torch.float32)
    y_t = torch.tensor(y_train, dtype=torch.long)
    bs  = max(2, min(batch_size, len(X_t)))
    dl  = DataLoader(TensorDataset(X_t, y_t), batch_size=bs,
                     shuffle=True, drop_last=len(X_t) > bs,
                     num_workers=0, pin_memory=torch.cuda.is_available(),
                     generator=torch.Generator().manual_seed(seed))

    model.train()
    for _ in range(epochs):
        for xb, yb in dl:
            xb = xb.to(device, non_blocking=True)
            yb = yb.to(device, non_blocking=True)
            opt.zero_grad()
            xr, mu, lv, z, logits = model(xb)
            loss = svae_loss(xr, xb, mu, lv, logits, yb,
                             beta, ce_weight, num_classes)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
            opt.step()
        sched.step()
    return model


@torch.no_grad()
def svae_encode(model, X):
    model.eval()
    X_t = torch.tensor(X, dtype=torch.float32).to(device)
    mu, _ = model.encode(X_t)
    return mu.cpu().numpy()


@torch.no_grad()
def svae_predict_proba_tta(model, X, n_rounds=TTA_ROUNDS):
    model.eval()
    X_t = torch.tensor(X, dtype=torch.float32).to(device)
    mu, logvar = model.encode(X_t)
    std = torch.exp(0.5 * logvar)

    accum = torch.zeros((len(X_t), model.classifier[-1].out_features),
                        device=device)
    for _ in range(n_rounds):
        eps = torch.randn_like(std)
        z   = mu + eps * std
        logits = model.classify(z)
        accum += F.softmax(logits, dim=1)
    return (accum / n_rounds).cpu().numpy()


def svae_predict_proba(model, X):
    model.eval()
    X_t = torch.tensor(X, dtype=torch.float32).to(device)
    mu, _ = model.encode(X_t)
    logits = model.classify(mu)
    return F.softmax(logits, dim=1).cpu().numpy()


def _fit_one_tabnet(Z_tr, y_tr, Z_va, y_va, n_classes, seed, weights=1):
    n_d_cfg = min(max(TABNET_N_D, Z_tr.shape[1] // 2), 64)
    n_a_cfg = min(max(TABNET_N_A, Z_tr.shape[1] // 2), 64)

    tab = TabNetClassifier(
        n_d=n_d_cfg, n_a=n_a_cfg,
        n_steps=TABNET_N_STEPS,
        gamma=TABNET_GAMMA,
        n_independent=1, n_shared=1,
        momentum=0.3, mask_type='entmax',
        verbose=0, seed=seed,
        device_name=TABNET_DEVICE,
        optimizer_params=dict(lr=2e-2),
    )
    with contextlib.redirect_stdout(io.StringIO()):
        tab.fit(Z_tr, y_tr,
                eval_set=[(Z_va, y_va)], eval_metric=['logloss'],
                max_epochs=TABNET_EPOCHS, patience=TABNET_PATIENCE,
                batch_size=max(8, min(TABNET_BATCH, len(Z_tr))),
                virtual_batch_size=max(4, min(TABNET_VBATCH, len(Z_tr))),
                weights=weights,
                drop_last=False)
    proba = tab.predict_proba(Z_va)
    full = np.zeros((len(Z_va), n_classes))
    for j, cls in enumerate(tab.classes_):
        full[:, int(cls)] = proba[:, j]
    return full


def run_trans_vatabnet(X_data, y_data, latent_dim=LATENT_DIM,
                       hidden_dim=HIDDEN_DIM, vae_epochs=VAE_EPOCHS,
                       tabnet_epochs=TABNET_EPOCHS, n_splits=N_SPLITS,
                       use_smote=True, beta=BETA_KL, ce_weight=CE_WEIGHT,
                       tabnet_seeds=TABNET_SEEDS, use_tta=True):
    n_samples, n_features = X_data.shape
    n_classes = len(np.unique(y_data))
    if n_classes < 2 or min(np.bincount(y_data)) < 4:
        return None

    counts = np.bincount(y_data)
    imbalance = counts.max() / counts.min()

    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
    oof_tab = np.zeros((n_samples, n_classes))
    oof_vae = np.zeros((n_samples, n_classes))

    for fold, (tr, va) in enumerate(skf.split(X_data, y_data)):
        X_tr, X_va = X_data[tr], X_data[va]
        y_tr, y_va = y_data[tr], y_data[va]

        scaler = RobustScaler()
        X_tr_s = scaler.fit_transform(X_tr)
        X_va_s = scaler.transform(X_va)

        vae_seed = 42 + fold * 100
        vae = train_supervised_beta_vae(
            X_tr_s, y_tr, num_classes=n_classes,
            hidden_dim=hidden_dim, latent_dim=latent_dim,
            epochs=vae_epochs, beta=beta, ce_weight=ce_weight,
            seed=vae_seed,
        )
        Z_tr = svae_encode(vae, X_tr_s)
        Z_va = svae_encode(vae, X_va_s)

        if use_tta:
            oof_vae[va] = svae_predict_proba_tta(vae, X_va_s, n_rounds=TTA_ROUNDS)
        else:
            oof_vae[va] = svae_predict_proba(vae, X_va_s)

        Z_tr_used = Z_tr.copy(); y_tr_used = y_tr.copy()
        if use_smote and imbalance >= 1.3:
            k = min(3, np.bincount(y_tr).min() - 1)
            if k >= 1:
                try:
                    Z_tr_used, y_tr_used = SMOTE(
                        k_neighbors=k, random_state=42
                    ).fit_resample(Z_tr, y_tr)
                except Exception:
                    pass

        seed_probas = []
        for s in tabnet_seeds:
            p = _fit_one_tabnet(Z_tr_used, y_tr_used, Z_va, y_va,
                                n_classes, seed=s + fold * 10, weights=1)
            seed_probas.append(p)
        oof_tab[va] = np.mean(seed_probas, axis=0)

    blend = 0.85 * oof_tab + 0.15 * oof_vae
    return blend


def best_accuracy(y_true, proba):
    if proba.ndim == 1:
        thr_grid = np.linspace(0.1, 0.9, 81)
        accs = [accuracy_score(y_true, (proba > t).astype(int)) for t in thr_grid]
        i = int(np.argmax(accs))
        thr = float(thr_grid[i])
        pred = (proba > thr).astype(int)
        return float(accs[i]), thr, pred
    else:
        pred = np.argmax(proba, axis=1)
        return float(accuracy_score(y_true, pred)), None, pred


# =============================================================================
# BINARY TASKS
# =============================================================================
results = []
conf_store = {}

def record_task(task_name, y_true, proba, class_names):
    if proba.ndim == 1:
        acc, thr, pred = best_accuracy(y_true, proba)
        auc = roc_auc_score(y_true, proba)
    else:
        acc, thr, pred = best_accuracy(y_true, proba)
        auc = roc_auc_score(y_true, proba, multi_class='ovr', average='macro')
    results.append({
        'Task':         task_name,
        'N':            len(y_true),
        'Classes':      len(class_names),
        'Accuracy':     round(acc, 4),
        'Balanced Acc': round(balanced_accuracy_score(y_true, pred), 4),
        'F1 (macro)':   round(f1_score(y_true, pred, average='macro'), 4),
        'ROC-AUC':      round(auc, 4),
        'MCC':          round(matthews_corrcoef(y_true, pred), 4),
    })
    conf_store[task_name] = {
        'y_true': y_true, 'y_pred': pred,
        'class_names': list(class_names),
        'cm': confusion_matrix(y_true, pred),
        'threshold': thr,
    }

print("\n" + "="*94)
print("  BINARY TASKS")
print("="*94)
print("[Task] Control vs PD")
proba = run_trans_vatabnet(X_A, y_bin)
record_task('Control vs PD', y_bin, proba[:, 1], le_bin.classes_)

print("\n[Task] Control vs LatePD")
proba = run_trans_vatabnet(X_A_late, y_late)
record_task('Control vs LatePD', y_late, proba[:, 1], np.array(['Control', 'Late']))

print("\n[Task] Control vs EarlyPD")
proba = run_trans_vatabnet(X_A_early, y_early)
record_task('Control vs EarlyPD', y_early, proba[:, 1], np.array(['Control', 'Early']))

# =============================================================================
# 3-CLASS STRATEGIES
# =============================================================================
print("\n" + "="*94)
print("  3-CLASS STRATEGIES")
print("="*94)

# ---- Base 3-class ----
print("\n[1] Base 3-class")
proba_3_base = run_trans_vatabnet(X_3, y_3)
acc_base = accuracy_score(y_3, np.argmax(proba_3_base, axis=1))
print(f"  → Acc = {acc_base:.4f}")

# ---- Multi-seed Early-vs-Late ----
print(f"\n[Support] Multi-seed Early-vs-Late ({N_EL_SEEDS} seeds)")
EL_seeds = TABNET_SEEDS[:N_EL_SEEDS]
proba_EL = run_trans_vatabnet(X_EL, y_EL, tabnet_seeds=EL_seeds)
acc_el = accuracy_score(y_EL, np.argmax(proba_EL, axis=1))
print(f"  → E-vs-L Acc = {acc_el:.4f}")

# ---- Control-vs-PD on full 3-class feature set (for hier) ----
print("\n[Support] Control-vs-PD on 3-class features")
proba_CvPD_full = run_trans_vatabnet(X_3, (y_3 > 0).astype(int))
acc_cvpd = accuracy_score((y_3 > 0).astype(int), np.argmax(proba_CvPD_full, axis=1))
print(f"  → C-vs-PD Acc = {acc_cvpd:.4f}")

# ---- Calibration of sub-probabilities ----
def calibrate_binary_oof(raw_proba, y_true, n_splits=N_SPLITS):
    if not USE_CALIBRATION:
        return raw_proba
    try:
        skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
        calib = np.zeros_like(raw_proba)
        for tr, va in skf.split(raw_proba, y_true):
            c = CalibratedClassifierCV(
                LogisticRegression(max_iter=2000),
                method='isotonic', cv=3)
            X_p = np.log(np.clip(raw_proba, 1e-6, 1-1e-6) /
                         (1 - np.clip(raw_proba, 1e-6, 1-1e-6))).reshape(-1,1)
            c.fit(X_p[tr], y_true[tr])
            calib[va] = c.predict_proba(X_p[va])[:,1]
        return calib
    except Exception as e:
        print(f"  [calibration] failed: {e}; using raw")
        return raw_proba

p_PD = proba_CvPD_full[:, 1]
p_Late_given_PD = np.zeros(len(y_3)); pd_mask = (y_3 > 0)
p_Late_given_PD[pd_mask] = proba_EL[:, 1]
p_Late_given_PD[~pd_mask] = 0.5

try:
    pd_idx = np.where(pd_mask)[0]
    calibrated_el = calibrate_binary_oof(proba_EL[:,1], y_EL)
    p_Late_given_PD[pd_idx] = calibrated_el
    print(f"  [calibration] Early-vs-Late calibrated")
except Exception as e:
    print(f"  [calibration] skipped: {e}")

# ---- Hierarchical ----
print("\n[2] Hierarchical (calibrated)")
proba_hier = np.stack([
    1.0 - p_PD,
    p_PD * (1.0 - p_Late_given_PD),
    p_PD * p_Late_given_PD,
], axis=1)
proba_hier = proba_hier / proba_hier.sum(axis=1, keepdims=True)
acc_hier = accuracy_score(y_3, np.argmax(proba_hier, axis=1))
print(f"  → Acc = {acc_hier:.4f}")

# ---- Stacking ----
def build_pairwise_oof(X, y_3, n_splits=N_SPLITS, seed=42):
    n_samples = len(y_3)
    pairs = [(0, 1), (0, 2), (1, 2)]
    oof_pairs = np.zeros((n_samples, len(pairs)))
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    for fold, (tr, va) in enumerate(skf.split(X, y_3)):
        for p_idx, (ci, cj) in enumerate(pairs):
            mask_tr = np.isin(y_3[tr], [ci, cj])
            idx_tr  = tr[mask_tr]
            y_tr_pair = (y_3[idx_tr] == cj).astype(int)
            if len(np.unique(y_tr_pair)) < 2:
                oof_pairs[va, p_idx] = 0.5; continue
            sc = RobustScaler()
            X_tr_s = sc.fit_transform(X[idx_tr])
            X_va_s = sc.transform(X[va])
            clf = LogisticRegression(C=1.0, penalty='l2', solver='lbfgs',
                                     max_iter=5000, class_weight='balanced',
                                     random_state=42)
            clf.fit(X_tr_s, y_tr_pair)
            oof_pairs[va, p_idx] = clf.predict_proba(X_va_s)[:,1]
    return oof_pairs

print("\n[3] Pairwise stacking (reference)")
pw_oof = build_pairwise_oof(X_3, y_3, n_splits=N_SPLITS)
skb = SelectKBest(f_classif, k=min(STACK_META_K, X_3.shape[1]))
X_top = skb.fit_transform(X_3, y_3)
X_meta = np.hstack([proba_3_base, pw_oof, X_top])

rskf = RepeatedStratifiedKFold(n_splits=N_SPLITS, n_repeats=N_STACK_REPEATS,
                               random_state=42)
meta_oof = np.zeros((len(y_3), 3)); hits = np.zeros(len(y_3))
for tr, va in rskf.split(X_meta, y_3):
    sc = StandardScaler()
    X_tr_s = sc.fit_transform(X_meta[tr])
    X_va_s = sc.transform(X_meta[va])
    clf = LogisticRegression(C=0.3, penalty='l2', solver='lbfgs',
                             max_iter=5000, class_weight='balanced',
                             multi_class='multinomial', random_state=42)
    clf.fit(X_tr_s, y_3[tr])
    meta_oof[va] += clf.predict_proba(X_va_s)
    hits[va] += 1
meta_oof /= np.maximum(hits[:,None], 1)
acc_stack = accuracy_score(y_3, np.argmax(meta_oof, axis=1))
print(f"  → Acc = {acc_stack:.4f}")

# =============================================================================
# SOFT BLEND (optimised on OOF)
# =============================================================================
print("\n[4] Soft blend (hier + stack + base)")
if USE_SOFT_BLEND:
    P = np.stack([proba_hier, meta_oof, proba_3_base], axis=0)

    def neg_acc(w):
        w = np.abs(w); w = w / (w.sum() + 1e-8)
        blend = np.tensordot(w, P, axes=1)
        return -accuracy_score(y_3, np.argmax(blend, axis=1))

    best_w = None; best_score = 1e9
    for start in [[0.6,0.3,0.1],[0.5,0.3,0.2],[0.4,0.4,0.2],
                  [0.7,0.2,0.1],[0.33,0.33,0.34]]:
        res = minimize(neg_acc, x0=np.array(start), method='Nelder-Mead',
                       options={'xatol':1e-4,'fatol':1e-6,'maxiter':400})
        if res.fun < best_score:
            best_score = res.fun; best_w = np.abs(res.x) / np.abs(res.x).sum()
    proba_blend = np.tensordot(best_w, P, axes=1)
    acc_blend = -best_score
    print(f"  → weights [hier, stack, base] = "
          f"[{best_w[0]:.3f}, {best_w[1]:.3f}, {best_w[2]:.3f}]")
    print(f"  → Acc = {acc_blend:.4f}")
else:
    proba_blend = proba_hier
    acc_blend = acc_hier

# =============================================================================
# PICK WINNER
# =============================================================================
strategies = {
    'base':       (acc_base,   proba_3_base),
    'hier':       (acc_hier,   proba_hier),
    'stack':      (acc_stack,  meta_oof),
    'soft_blend': (acc_blend,  proba_blend),
}

print("\n" + "-"*94)
print("  3-CLASS STRATEGY COMPARISON")
print("-"*94)
for name, (acc, _) in sorted(strategies.items(), key=lambda x: -x[1][0]):
    print(f"  {name:<14}  Acc = {acc:.4f}")

best_name = max(strategies, key=lambda k: strategies[k][0])
best_acc, best_proba = strategies[best_name]
print(f"\n  → WINNER: {best_name}  (Acc = {best_acc:.4f})")

record_task(f'3-class ({best_name})', y_3, best_proba, le_3.classes_)

# =============================================================================
# OVERALL TABLE
# =============================================================================
print("\n" + "="*94)
print("    OVERALL PERFORMANCE — Trans-VaTabNet")
print(f"    latent={LATENT_DIM}  hidden={HIDDEN_DIM}  β={BETA_KL}  "
      f"λ={CE_WEIGHT}  dropout={DROPOUT}")
print(f"    {N_SPLITS}-fold  {len(TABNET_SEEDS)}-seed  TTA={TTA_ROUNDS}  "
      f"EL_seeds={N_EL_SEEDS}")
print("="*94)
df_overall = pd.DataFrame(results)
print(df_overall.to_string(index=False))

print("\n" + "-"*94)
print(f"  BEST SINGLE TASK  : "
      f"{df_overall.loc[df_overall['Accuracy'].idxmax(),'Task']}  "
      f"(Acc = {df_overall['Accuracy'].max():.4f})")
print(f"  MEAN ACCURACY     : {df_overall['Accuracy'].mean():.4f}")
print(f"  MEAN ROC-AUC      : {df_overall['ROC-AUC'].mean():.4f}")
print(f"  MEAN F1 (macro)   : {df_overall['F1 (macro)'].mean():.4f}")
print(f"  MEAN MCC          : {df_overall['MCC'].mean():.4f}")
print("="*94)

# =============================================================================
# CONFUSION MATRICES
# =============================================================================
def show_confusion_matrix(task_name, cm, class_names):
    n = len(class_names)
    short = [str(c)[:10] for c in class_names]
    col_w = max(10, max(len(s) for s in short) + 2)
    print("\n" + "="*94)
    print(f"  CONFUSION MATRIX — {task_name}")
    print("="*94)
    header = " "*20 + "".join(f"{'pred:'+s:>{col_w}}" for s in short) + f"{'total':>{col_w}}"
    print(header); print("-" * len(header))
    for i, true_name in enumerate(short):
        row_total = cm[i].sum()
        cells = "".join(f"{cm[i, j]:>{col_w}d}" for j in range(n))
        print(f"{'true:'+true_name:<20}{cells}{row_total:>{col_w}d}")
    col_tot = cm.sum(axis=0)
    print("-" * len(header))
    print(f"{'TOTAL':<20}" + "".join(f"{col_tot[j]:>{col_w}d}" for j in range(n)))
    print("\n  Per-class accuracy (recall) and precision:")
    for i, name in enumerate(short):
        denom = cm[i].sum()
        rec = cm[i, i] / denom if denom else 0.0
        prec_denom = cm[:, i].sum()
        prec = cm[i, i] / prec_denom if prec_denom else 0.0
        print(f"    {name:<12}  recall = {rec:.4f}   precision = {prec:.4f}   "
              f"(support = {denom})")
    print("\n  Row-normalised matrix:")
    print(" "*20 + "".join(f"{'pred:'+s:>{col_w}}" for s in short))
    for i, true_name in enumerate(short):
        denom = cm[i].sum() or 1
        cells = "".join(f"{cm[i, j]/denom:>{col_w}.3f}" for j in range(n))
        print(f"{'true:'+true_name:<20}{cells}")

for task_name in ['Control vs PD', 'Control vs LatePD', 'Control vs EarlyPD',
                  f'3-class ({best_name})']:
    if task_name in conf_store:
        rec = conf_store[task_name]
        show_confusion_matrix(task_name, rec['cm'], rec['class_names'])

# =============================================================================
# SHAP
# =============================================================================
print("\n" + "="*94)
print("  SHAP INTERPRETABILITY")
print("="*94)

gene_names_3 = feature_set_3.copy()
ensg_map_feat = ensg_to_symbol_map(gene_names_3)
readable_names = [ensg_map_feat.get(g, g) for g in gene_names_3]

scaler_shap = StandardScaler()
X_3_std     = scaler_shap.fit_transform(X_3)

shap_model = LogisticRegression(
    C=0.5, penalty='l2', solver='lbfgs', max_iter=5000,
    class_weight='balanced', multi_class='multinomial', random_state=42)
shap_model.fit(X_3_std, y_3)

background = shap.sample(X_3_std, min(50, len(X_3_std)), random_state=42)
explainer  = shap.KernelExplainer(
    lambda x: shap_model.predict_proba(x), background, link="logit")

n_explain = min(100, len(X_3_std))
X_explain = X_3_std[:n_explain]

print(f"[shap] KernelSHAP for {n_explain} × {X_3_std.shape[1]}…")
shap_values = explainer.shap_values(X_explain, nsamples=100)


def _normalise_shap(sv_in, n_classes, n_feats, n_samples):
    arr = np.asarray(sv_in, dtype=np.float64)
    if arr.ndim == 3:
        if arr.shape == (n_classes, n_samples, n_feats):
            return [arr[c] for c in range(n_classes)]
        if arr.shape == (n_samples, n_feats, n_classes):
            return [arr[:, :, c] for c in range(n_classes)]
        if arr.shape == (n_classes, n_feats, n_samples):
            return [arr[c].T for c in range(n_classes)]
        raise ValueError(f"[shap] 3D shape {arr.shape}")
    if arr.ndim == 2:
        if arr.shape == (n_samples, n_feats):
            return [arr]
        if arr.shape[0] == n_classes * n_samples and arr.shape[1] == n_feats:
            return [arr[c*n_samples:(c+1)*n_samples] for c in range(n_classes)]
        raise ValueError(f"[shap] 2D shape {arr.shape}")
    if arr.ndim == 4 and arr.shape[-1] == 1:
        arr = arr[..., 0]
        if arr.shape == (n_samples, n_feats, n_classes):
            return [arr[:, :, c] for c in range(n_classes)]
    raise ValueError(f"[shap] ndim={arr.ndim} shape={arr.shape}")


shap_per_class = _normalise_shap(shap_values, 3, X_3_std.shape[1], X_explain.shape[0])
print(f"[shap] normalised to {len(shap_per_class)} classes, each {shap_per_class[0].shape}")
while len(shap_per_class) < 3:
    shap_per_class.append(np.zeros_like(shap_per_class[0]))

mean_abs = np.stack([np.abs(s).mean(axis=0) for s in shap_per_class[:3]])
overall = mean_abs.mean(axis=0)

imp_df = pd.DataFrame({
    'feature':              readable_names,
    'ensg':                 gene_names_3,
    'mean_abs_shap':        overall,
    'shap_class_0_Control': mean_abs[0],
    'shap_class_1_Early':   mean_abs[1],
    'shap_class_2_Late':    mean_abs[2],
}).sort_values('mean_abs_shap', ascending=False)

imp_df.to_csv('shap_global_importance.csv', index=False)
print("\nTop-20 features by mean |SHAP|:")
print(imp_df.head(20)[['feature', 'mean_abs_shap']].to_string(index=False))

print("\nPaper-biomarker SHAP ranks:")
feat_to_rank = {n: i+1 for i, n in enumerate(imp_df['feature'].tolist())}
feat_to_shap = dict(zip(imp_df['feature'], imp_df['mean_abs_shap']))
for sym in biomarker_symbols:
    if sym in feat_to_rank:
        print(f"  {sym:<8}  rank = {feat_to_rank[sym]:>4}  "
              f"mean|SHAP| = {feat_to_shap[sym]:.5f}")
    else:
        print(f"  {sym:<8}  not in feature set")

top20 = imp_df.head(20).iloc[::-1]
plt.figure(figsize=(8, 7))
plt.barh(top20['feature'], top20['mean_abs_shap'])
plt.xlabel('mean |SHAP|')
plt.title('Top-20 genes by overall SHAP importance')
plt.tight_layout()
plt.savefig('shap_top20_bar.png', dpi=200)
plt.show()

try:
    shap.summary_plot(shap_per_class[2], X_explain,
                      feature_names=readable_names, max_display=20, show=False)
    plt.savefig('shap_beeswarm_late.png', dpi=200, bbox_inches='tight')
    plt.show()
except Exception as e:
    print(f"[shap] beeswarm skipped: {e}")

np.save('shap_values_class0.npy', shap_per_class[0])
np.save('shap_values_class1.npy', shap_per_class[1])
np.save('shap_values_class2.npy', shap_per_class[2])
np.save('shap_X_explain.npy',     X_explain)

# =============================================================================
# SAVE
# =============================================================================
df_overall.to_csv('overall_performance.csv', index=False)
acc_matrix = df_overall.set_index('Task')[
    ['Accuracy','Balanced Acc','F1 (macro)','ROC-AUC','MCC']]
acc_matrix.to_csv('accuracy_matrix.csv')

for task_name, rec in conf_store.items():
    safe = (task_name.replace(' ', '_').replace('/', '-')
                     .replace('(', '').replace(')', ''))
    pd.DataFrame(rec['cm'],
                 index=[f"true_{c}" for c in rec['class_names']],
                 columns=[f"pred_{c}" for c in rec['class_names']]
                 ).to_csv(f'confusion_matrix_{safe}.csv')

for key, res in enrich_results.items():
    if res is not None and hasattr(res, 'results') and res.results is not None:
        safe = key.replace('::', '_').replace(' ', '_')
        res.results.to_csv(f'enrichr_{safe}.csv', index=False)

np.save('oof_3class_winner.npy', best_proba)
np.save('oof_3class_hier.npy',   proba_hier)
np.save('oof_3class_stack.npy',  meta_oof)
np.save('oof_3class_base.npy',   proba_3_base)

# ---------- Shared / common DEG artifacts ----------
pd.Series(sorted(shared_sig)).to_csv('shared_DEGs_ensg.txt',
                                     index=False, header=False)
pd.Series(sorted(shared_all)).to_csv('shared_all3_DEGs_ensg.txt',
                                     index=False, header=False)
pd.Series(sorted(early_only)).to_csv('early_only_DEGs_ensg.txt',
                                     index=False, header=False)
pd.Series(sorted(late_only)).to_csv('late_only_DEGs_ensg.txt',
                                    index=False, header=False)
pd.Series(sorted(early_specific)).to_csv('early_specific_DEGs_ensg.txt',
                                         index=False, header=False)
pd.Series(sorted(late_specific)).to_csv('late_specific_DEGs_ensg.txt',
                                        index=False, header=False)
pd.Series(sorted(two_of_three)).to_csv('two_of_three_DEGs_ensg.txt',
                                       index=False, header=False)

with open('shared_DEGs_symbols.txt', 'w') as f:
    f.write("\n".join(to_symbols(shared_sig)))
with open('shared_all3_DEGs_symbols.txt', 'w') as f:
    f.write("\n".join(to_symbols(shared_all)))
with open('early_only_DEGs_symbols.txt', 'w') as f:
    f.write("\n".join(to_symbols(early_only)))
with open('late_only_DEGs_symbols.txt', 'w') as f:
    f.write("\n".join(to_symbols(late_only)))

# Shared-DEG summary table
shared_summary = pd.DataFrame({
    'Set': [
        'Early-specific',
        'Late-specific',
        'Late-vs-Early',
        'Shared (Early ∩ Late)',
        'Shared in ALL 3 contrasts',
        'Early-only',
        'Late-only',
        'Significant in 2-of-3',
        'Union of all DEGs',
    ],
    'N_genes': [
        len(early_specific),
        len(late_specific),
        len(le_sig),
        len(shared_sig),
        len(shared_all),
        len(early_only),
        len(late_only),
        len(two_of_three),
        len(all_sig),
    ],
})
shared_summary.to_csv('DEG_intersection_summary.csv', index=False)
print("\nDEG intersection summary:")
print(shared_summary.to_string(index=False))

print("\nSaved: overall_performance.csv, accuracy_matrix.csv")
print("Saved: confusion_matrix_*.csv, enrichr_*.csv")
print("Saved: shap_global_importance.csv, shap_top20_bar.png, shap_beeswarm_late.png")
print("Saved: oof_3class_*.npy (winner + variants)")
print("Saved: shared_DEGs_early_and_late.csv")
print("Saved: DEG_intersection_summary.csv")
print("Saved: shared_DEGs_ensg.txt / shared_DEGs_symbols.txt")
print("Saved: shared_all3_DEGs_ensg.txt / shared_all3_DEGs_symbols.txt")
print("Saved: early_only_DEGs_*.txt / late_only_DEGs_*.txt")
print("Saved: early_specific_DEGs_ensg.txt / late_specific_DEGs_ensg.txt")
print("Saved: two_of_three_DEGs_ensg.txt")
```
